#### I can change the value of an item in the list by resetting it. For example: my_list[1] = "new value". I can also use methods on lists. Some methods I can use are .append() and .insert(). Append adds an item to the end of the list. It takes one parameter, which is the value you want to add. Insert adds an item in a specific place in the list, shifting everything after it to the right. It takes two parameters, the location you want to insert the value and the value that should be inserted.

```python
activities = ["sleep", "eat", "run", "swim", "play", "clean"]
print(activities)

#activities.remove("clean")
activities.pop(5)
print(activities)
```

#### I have learned 2 new methods that can be applied to lists, .pop() and .remove(). Both can remove an item from a list, but .pop() requires the index while .remove() requires the exact value.

```python
numbers = [3, 5, 8, 10, 11, 28, 42, 77, 101, 108, 2002]
print(numbers)

#change 8 to 1
numbers[2] = 1
#insert 50 at index 2
numbers.insert(2, 50)
#remove 108 by value
numbers.remove(108)
#add 100 to the end
numbers.append(100)
#remove 2002 by negative index
numbers.pop(-2)
#remove 11 by postive index
numbers.pop(5)
#use delete to delete the 5
del(numbers[1])

print(numbers)
```

```python
cool = ["video", "cool", "ball", "cat", "cookie"]
cool.append("Hello")
print(cool)
cool.insert(3, 1)
print(cool)
del(cool[1])
print(cool)
cool.remove("Hello")
print(cool)
last_item = cool.pop()
print(f"The last word in my list was {last_item}.")
```

#### .pop() has three functionalities. The first one is to remove something by index. The second is to store the value it deletes so you can set it as a variable. The third is to automatically remove the last item if no index is provided.
