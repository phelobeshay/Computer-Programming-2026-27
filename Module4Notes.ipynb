```python
family_toppings = {
    "Dad" : "sausage",
    "Mom" : "mushrooms",
    "Brother" : "pepperoni",
    "Me" : "cheese"
}

family_toppings["Grandma"] = "berries"
family_toppings["Brother"] = "peppers"
print(family_toppings["Mom"])
print(len(family_toppings))
print(family_toppings)

print(family_toppings["Dad"].title())
print(family_toppings)

family_toppings["Dad"] = family_toppings["Dad"].title()
print(family_toppings)

#print(family_toppings.get("aliyah", "Person isn't present"))
#print(family_toppings.get("Dad", "Person isn't present"))

for k,v in family_toppings.items():
    print(f"{k} likes {v}")

for names in family_toppings.keys():
    print(names.title())
for toppings in family_toppings.values():
    print(toppings.title())
```

```python
census = {
    "household 1":{
        "name":"Blake",
        "age":100,
        "income":100000
    },
    "household 2":{
        "name":"Julien",
        "age":20,
        "income": [6.55, 503593, 59, 403]
    }
}

print(census["household 2"]["income"][2])

print(len(census))
print(len(census["household 2"]))
print(len(census["household 2"]["income"]))
```

#### Lists allow us to store many values in one place and access them by their index number. Dictionaries allow us to store things by key and we access the value by refrencing the key. It is important to keep track of what data type we are using at any given time. In the example above, census is a dictionary. census["household 2"] is also a dictionary. census["household 2"]["income"] is a list. census["household 2"]["income"][2] is an integer.
