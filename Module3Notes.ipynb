```python
x = 5
y = 5
z = 6

#comparison operators to build one statement
print(x!=y)
print(x!=z)
print(x==y)
print(x==z)
print(x>=y)
print(x>y)
print(x<y)
print(x<=y)

#logical operators to compare multiple statement that each were built with comparison operators
print(x!=y and x!=z)
print(x!=y or x!=z)
print(not(x!=y))
```

#### We can use an if statement to alter the flow of code. After the word if we include an expression that resolves to true or false. This expression can be built using comparison and logical operators. If we want one of two things to happen after our if, we can have an else statement. If there are multiple routes, we can include one or more elifs inbetween the if and else statements.

```python
age = input("What is your age? ")
age = int(age)

if age >= 16:
    print("go take your driving test lil bro")
elif age >= 35:
    print("how's your midlife crisis unc")
elif age >= 65:
    print("I'm old!")
else:
    print("lil bro")
```

```python
guess = int(input("Guess a number: "))
if guess >= 10:
    if guess <= 20:
        print("You are within range.")
    else:
        print("Guess is too large.")
else: 
    print("Guess is too small.")
```

```python
clothing_stores = ["walmart", "pandabuy", "nike", "peter millar", "dior"]
has_store = False
fav_store = input("Enter your favorite store: ")
for stores in clothing_stores:
    if stores == fav_store.lower():
        has_store = True

if has_store == True:
    print("We have that store.")
else:
    print("We don't have that store.")
```

```python
has_egg = False
pizza_toppings = ["cheese", "pepperoni", "sausage", "mushrooms", "sauce", "chicken"]

for topping in pizza_toppings:
    if topping == "eggs" or topping == "egg":
        has_egg = True

if has_egg == True:
    print("You get an F in the class.")
elif has_egg == False:
    print("You get an A in the class.")
```
